```python
import numpy as np

np.random.seed(42)
student_ids = np.arange(1, 201)
attendance = np.clip(np.random.normal(85, 10, 200), 50, 100).round(1)
study_hours = np.clip(np.random.normal(4, 2, 200), 0, 10).round(1)
assignments_completed = np.random.randint(0, 11, 200)
noise = np.random.normal(0, 8, 200)

marks = np.clip(
    40 + attendance * 0.3 + study_hours * 4 + assignments_completed * 1.5 +
    noise,
    0, 100
).round(1)
```

```python
marks.max()
```

```python
marks.min()
```

```python
marks.mean()
```

```python
np.median(marks)
```

```python
marks.std()
```

```python
marks.var()
```
